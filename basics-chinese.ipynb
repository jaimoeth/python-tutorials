# 1. 计算

```python
"""
面积差值：
如果圆形的半径和正方形边长都是2, 那么这个圆形的面积比正方形大了多少呢?
如果圆形的半径和正方形边长都是3呢?
"""
```

## 笔记

```python
pi = 3.14159                            # 给变量"赋值"

radius = 2
length = radius
circle_area = pi * (radius ** 2)        # "计算"面积 | +[加]、-[减]、*[乘]、/[除]、//[取整]、%[取余]、**[幂]
square_area = length ** 2
diff = circle_area - square_area
print("The difference is: ", diff)      # "输出"面积 | 输出中", "会输出为"空格"

radius = 3
length = radius
circle_area = pi * (radius ** 2)
square_area = length ** 2
diff = circle_area - square_area
print("The new difference is: ", diff)

"""
注意：
选中多行内容后使用快捷键 Ctrl+/ 可以快速注释或取消注释
在 VS Code 中，输入变量名的前几个字母后按 Tab键可以自动补全变量名
"""
```
